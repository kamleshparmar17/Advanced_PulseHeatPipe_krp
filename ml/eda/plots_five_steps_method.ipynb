```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import seaborn as sns
```

```python
df =  pd.read_excel("../../data/observation_alpha_beta.xlsx")
```

```python
df.head()
```

Plots for various Alpha and constant Beta (0°)

```python
# Set style and context for publication-quality visuals
sns.set_context("notebook", font_scale=1.2)
sns.set_style("whitegrid")

# Define a color palette for clear differentiation
palette = sns.color_palette("tab10")  # Use a vivid palette for clarity

# Create the plot
plt.figure(figsize=(8, 6))

# Plot the data with enhancements
sns.lineplot(
    data=df,
    x="alpha",
    y="TR1[K/W]",
    hue="Q[W]",
    style="Q[W]",  # Different line styles based on "Q[W]"
    markers=True,  # Add markers for data points
    dashes=False,  # Use solid lines
    palette=palette,  # Use the vivid color palette
    marker='o',  # Circular markers
    markerfacecolor='auto',  # Markers filled with line color
    markeredgecolor='black',  # Black edge for better contrast
    markeredgewidth=1  # Adjust the edge width
)

# Add title and labels
plt.title("Thermal Resistance (TR1) vs Alpha", fontsize=14, fontweight='bold')
plt.xlabel("Alpha [°]", fontsize=12)
plt.ylabel("Thermal Resistance (TR1) [K/W]", fontsize=12)

# Customize legend
plt.legend(
    title="Heat Input [Q(W)]",
    fontsize=10,
    title_fontsize=12,
    loc="upper left",  # Place the legend at the upper-right corner
    frameon=True,  # Add a box around the legend
    shadow=False   # Avoid shadow for a cleaner look
)

# Add grid lines for better readability
plt.grid(True, which='both', linestyle='--', linewidth=0.6, alpha=0.8)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
# Filter the data for alpha only (example: alpha == 90)
alpha_only_data = df[df["alpha"] == 90]  # Adjust the filter as needed

# Plot the graph
plt.figure(figsize=(8, 6))
sns.lineplot(
    data=alpha_only_data,
    x="Q[W]",
    y="Te1[K]",
    marker="o",
    label="Te1[K]",
    linewidth=1,
    color='blue'
)
sns.lineplot(
    data=alpha_only_data,
    x="Q[W]",
    y="Te2[K]",
    marker="o",
    label="Te2[K]",
    linewidth=1,
    color='red'
)
sns.lineplot(
    data=alpha_only_data,
    x="Q[W]",
    y="Tc1[K]",
    marker="o",
    label="Tc1[K]",
    linewidth=1,
    color='green'
)
sns.lineplot(
    data=alpha_only_data,
    x="Q[W]",
    y="Tc2[K]",
    marker="o",
    label="Tc2[K]",
    linewidth=1,
    color='orange'
)

# Enhance the graph for publication
plt.title("Temperature vs Heat Input at Alpha 90° and Beta 0°", fontsize=14)
plt.xlabel("Heat Input Q [W]", fontsize=12)
plt.ylabel("Temperature [K]", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=10, title="Temperature Metrics")
plt.tight_layout()

# Show the plot
plt.show()
```
